```python
#closure
def outer():
    message = "Hi"
    def inner():
        print(message)
    return inner

greet = outer()
greet()
print("inspect closure", greet)
print("inspect using closure", greet.__closure__)
print("inspect cell contents", greet.__closure__[0].cell_contents)
```

```python
#use of nested functions
def process_marks(marks):
    def is_valid_mark(mark):
        return 0 <= mark <= 100
    for mark in marks:
        if is_valid_mark(mark):
           print(mark)

process_marks([90, 85, 110, -5, 75])
```

```python
#student marks grade calculation using closure
def create_grade_calculator(pass_mark):
    def calculate_grade(marks):
        if marks >=pass_mark:
            return "Pass"
        else:
            return "Fail"
    return calculate_grade

result_checker=create_grade_calculator(40) #closure function
ug_checker=create_grade_calculator(40) #closure function
pg_checker=create_grade_calculator(50) #closure function

print("Raju: ", result_checker(45)) #Pass
print("Sita: ", result_checker(35)) #Fail
```

```python
#why non local is required

def create_account(initial_balance):
    balance = initial_balance

    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance
    return deposit
account=create_account(5000)
print(account(1000)) #5500
print(account(500)) #1500


```
